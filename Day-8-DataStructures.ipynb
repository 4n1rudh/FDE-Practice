## Strings - String Indexing, String Slicing, String Methods, String Immutability

### String Indexing
Indexing: We can call an individual letter from a word

            Positive Index starts from 0 towards right
            
            Negative Index starts at -1 from the right to the left

            
P  Y  T  H  O  N

0  1  2  3  4  5

-6 -5 -4 -3 -2 -1

```python
#Postive Index
word = "PYTHON"
print(word[0])
print(word[3])
```

```python
#Negative Index
word = "PYTHON"
print(word[-1])
print(word[-4])
```

```python
fruits = ["orange","grape","banana"]
print(fruits[2])
```

### String Slicing
Slicing: Extracting a particular portion of String, List or Tuple etc.

Start Parameter: Index where slicing begins, also is incuded
Stop Parameter: Index where slicing ends, also excluded
Step Parameter: Number of positions to shift
        
        Positive Step: Start should be less than the stop (Left to Right) incremental slicing
        
        Negative Step: Start should be greater than the stop (Right to Left) decremental slicing

```python
word = "PYTHON"
print(word[3:5])
#start is 3, so it's included
#stop is 4, so it's excluded
```

```python
word = "PYTHON"
print(word[5:3])
#start is 5; it is greater than the end
#stop is 3; it is less than the start
#so the print won't happen
```

```python
word = "PYTHON"
print(word[2:7])
#start is 2; it is included 
# stop is 7; but the final index is 5, there is no 7. So it will pring till whatever is the last index.
```

```python
# Omitting the start index defaults the start index to 0
word = "PYTHON"
print(word[:3])
```

```python
# Omitting the stop index defaults the stop index to the end
word = "PYTHON"
print(word[3:])
print(word[:]) # If neither is given, the full word is printed
```

```python
word = "PYTHON"
print(word[-4:3])
```

```python
word = "PYTHON"
print(word[-5:-4])
```

```python
word = "PYTHON"
print(word[-4:-5]) # No output because, without giving a step, step defaults to 1. Step decides the direction.
print(word[-4:-5:-1]) #now the negative step rule applies, so output will be shown
```
