## List Immutability
### Characteristics: Index, Slicing, Mutable

```python
list = ["apple", "banana", "mango", "orange", "grapes"]
print(list)
```

```python
print(list[4]) #Calling position by using index
```

```python
print(list[1:4])
```

## Mutable - add, update, remove, insert, extend, sort

```python
#update
fruits = ["apple", "banana", "orange"]
fruits[0]="mango"
print(fruits)
```

```python
#By default, append adds a new element to the end of the list
fruits = ["apple", "banana", "orange"]
fruits.append("mango")
print(fruits)
```

```python
#Append with multiple elements
fruits = ["apple", "banana", "orange"]
fruits.append(["mango", "grape", "cherry"]) #Python adds the multi-element list as a single element
print(fruits)
```

```python
#use extend instead of append to add as individual elements
fruits = ["apple", "banana", "orange"]
fruits.extend(["mango", "grape", "cherry"]) 
print(fruits)
```

```python
# In the insert method, you can insert at any position you want
fruits = ["apple", "banana", "orange"]
fruits.insert(1, "mango") 
print(fruits)
```

```python
# Remove, removes an element by value
fruits = ["apple", "banana", "orange"]
fruits.remove("banana") 
print(fruits)
```

```python
#We can use delete method to delete specific element by calling index
fruits = ["apple", "banana", "orange"]
del fruits[1]
print(fruits)
```

```python
# Pop method: can remove a specific element by index
fruits = ["apple", "banana", "orange"]
fruits.pop(1)
print(fruits)
```

```python
# Sort method
fruits = ["banana", "orange", "apple"]
fruits.sort()
print(fruits)
```

## Tuple supports indexing and slicing, it's immutable, so won't support add, update, remove, insert, extend and sort

## Sets {}
### Stores unique values only and doesn't allow duplicates

```python
numbers = {10,20,30,40,50}
print(numbers)
#output order may vary
```

```python
numbers = {10,60,40,10,20,30,40,50}
print(numbers)
#output order may vary, removes duplicates
```

```python
numbers = {10,20,30,40,50}
value = numbers.pop()
print(value)
```

## Dictionary
### Stores data in key-value pairs

```python
#Allows duplicates but overwrites the old value to latest one
students = {
    "name":"anirudh",
    "course":"FDE",
    "age": 35
}
print(students)
print(students["name"])
```

```python
#adding in a dictionary. Similar to update
students = {
    "name":"anirudh",
    "course":"FDE",
    "age": 35
}
students["city"]="hyd"
print(students)
```
