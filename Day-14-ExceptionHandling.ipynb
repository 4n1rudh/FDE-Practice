## Python Exception Handling
### Try, Exception, Else, Finally, Raise etc.

```python
try:
    a=10
    b=0
    print(a/b)
except:
    print("Can't be divided by 0")
```

```python
try:
    a=10
    b=0
    print(a/b)
except Exception as e: #e is the variable that stores the exception object (error information)
    print(e)
```

```python
count=1
while count<=3:
    try:
        a=10
        b=1
        print(a/b)
        count=+1
    except Exception as e:
        print(e)
```
