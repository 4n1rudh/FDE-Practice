## Left Shift

```python
x=15
x<<=1
print(x)

# Binary representation of 15 is 1111
# Left Shift 1 = shifting 1111 to the left by adding 0 = 11110 = 30
```

```python
x=19
x<<=2
print(x)

# Binary representation of 19 is 10011
# Left Shift 2 = shifting 10011 to the left by adding two 0s = 101100 = 76
```

```python
x=21
x<<=3
print(x)
```

## Right Shift

```python
x=15
x>>=1
print(x)

# Binary representation of 15 is 1111
# Right Shift 1 = shifting 1111 to the right by adding 0 first and eliminating the last = 0111 = 7
```

## Bitwise Operators (AND OR XOR NOT)

```python
a=5 #0101
b=3 #0011
print("Bitwise AND:", a & b) #& is true only if both are true, 0001
```

```python
a=5
b=3
print("Bitwise OR:", a | b) #| is true if any one true, 0111
```

```python
a=5
b=3
print("Bitwise XOR:", a ^ b) #^ is false if both are same; either true or false, 0110
```

```python
a=5
print("Bitwise NOT:", ~a)
#~a = -(a+1)
```
