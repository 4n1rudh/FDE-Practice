## While Loop
### While loop is used to repeat block of code as long as the given condition is true
### While loop can use index (i) to access a list of elements
### For loop is used to iterate over the sequence of elements (list, tuple, dictionary, sets etc.)

```python
#Infinity Loop
#while True:
#    print("hello")
```

```python
i=1
while i<=1:
    print(i)
    i=i+1
```

```python
i=1
while i in range (1,4):
    password = input("Enter your password")
    if password == "akki":
        print("You are logged in")
        break
    else:
        print("Incorrect password")
    i=i+1
```

```python
#for loop example
for i in range (1,4):
    question=input("Ask AI")
    if question.lower() == "exit":
        print("chatbot stopped")
        break
    print("AI is responding")
```
