## Python Lambda Function
### Lambda Function is a small anonymous function (a function without a formal name)

```python
#Syntax for Lambda Function
lambda arguments : expression
```

```python
def add (a, b):
    return(a+b)
result = add(10, 20)
print(result)
```

```python
add = lambda a, b: a+b #Here a, b are parameters, a+b is expression. add is a variable that stores a lambda function. lambda is a keyword
result = add(10, 20) #arguements 10,20
print(result)
#Note: lambda doesn't need a return keyword; it automatically returns the expression result.
```

```python
anirudh = lambda x: x*x
print(anirudh(5))
```

```python
check = lambda x : "even" if x % 2 == 0 else "odd"
print(check(11))
```

```python
check = lambda x : "positive" if x > 0 == 0 else "negative"
print(check(11))
print(check(-11))
```

```python
num = eval(input("enter number"))
if num>10:
    print("number is greater than 10")
else:
    print("number is less than 10")
```

## Walrus Operator
### Walrus Operator is used to value to variable use same variable in same expression

```python
#Without Walrus Operator
numbers=[10,20,30,40]
count = len(numbers)
if count >3:
    print(count)
```

```python
#With Walrus Operator
numbers = [10,20,30,40]
count = len(numbers)
if (count:= len(numbers)> 3):
    print(count)
```

```python
if (num := eval(input("enter number")))>10:
    print("number is greater than 10")
else:
    print("number is less than 10")
```

```python
lambda x:x > 10
if (num := eval(input("enter number"))):
    print ("number is greater than 10")
else:
    print("number is less than 10")
```

```python
def ani(x):
    a=x*2
    b=x/2
    c=a+b
    return c
print(ani(5))
```

```python
## lambda doesn't support multiple assignments like a =, b = etc.
ani = lambda x : (
    a=x*2
    b=x/2
    c=a+b )
    return c
print(ani(5))
```

```python
ani = lambda x : (x*2) + (x/2)
print(ani(5))
```
