## Type Conversion

```python
#Converting from one data type to another

a=2
b=3.5
c=a+b
print(c)
c=int(c)
print(c)
print(type(c))
```

```python
age="25"
age= int(age)
print(age)
print(type(age))
```

```python
age="25"
age=set(age)
print(age)
```

```python
age="25"
age=tuple(age)
print(age)
```

```python
#Example
price = float(input("Enter product price")) #input always returns string
tax = price * 0.18
Total = price + tax
print(Total)
```

## FString (Format String)

```python
#introduced in Python 3.6
name = "chitti"
print("hello", name)
```

```python
name = "chitti"
print(f"hello {name}")
```

```python
name =  input("Enter your name:")
age = input("Enter your age:")
weight = input("Enter your weight:")
print("His name is " + name + " His age is " +age+ " His weight is " +weight)
print("His name is",name,"His age is",age,"His weight is",weight)
print(f"His name is {name} His age is {age} His weight is {weight}")
```
