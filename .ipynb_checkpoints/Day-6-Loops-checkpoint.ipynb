```python
#for loop we can iterate over a sequesnce of items, list, tuple, range, dictionary etc.
#for loop with string example
a="python"
for i in a:
    print(i)
```

```python
#for loop with list example
number = ["azure", "aws", "gcp"]
for i in number:
    print(i)
```
