## Conditions if, elif, else, nested condition

### Rule 1: use colon (:) every if, else, elif
### Rule 2: Indentation is mandatory

```python
#simple if statement
age=20
if age>=18:
    print("You are an adult")
else:
    print("You are a minor")
```

```python
# Pass value dynamically using input function
age=input("Enter age")
if int(age) >= 18:
    print("You are a major")
else:
    print("You are a minor")
```

### if and elif

```python
marks=89
if marks>= 90:
    print("Grade A")
if marks>=80:
    print("Grade B")
if marks>=70:
    print("Grade C")
else:
    print("Fail")
#See that multiple grades are being displayed. So, use elif.
```

```python
marks=89
if marks>= 90:
    print("Grade A")
elif marks>=80:
    print("Grade B")
elif marks>=70:
    print("Grade C")
else:
    print("Fail")
```

## Nested Conditions
### Condition inside the condition

```python
marks=70
lab=80
if marks>=60:
    print("Pass")
    if lab>=80:
        print("Eligible for 1st grade")
else:
    print("Fail")
```
